```python
with open ("stopwords Japanese.txt", encoding='utf-8') as f:
    stopwords = f.read().split('\n')
df = pd.read_csv('hokkyoku_1949_2025.csv', index_col = False)
```

```python
df = replace_jap_names(df)
```

```python
corpus_analyzer_2022 = SpeechCharacteristics(df,2022,2025)
speeches_2022 = corpus_analyzer_2022.count_speech_number()
speeches_2022.to_csv('tables/speech_number_df_2022-2025.csv', index=False)
```

```python
invited_speakers = corpus_analyzer_2022.get_share_of_guest_speakers()
invited_speakers.to_csv('tables/invited_speakers_2022-2025.csv', index=False)
invited_speakers
```

```python
party_membership = corpus_analyzer_2022.get_party_shares(14,14)
party_membership.to_csv('tables/party_membership_2022-2025.csv', index=False)
party_membership
```
