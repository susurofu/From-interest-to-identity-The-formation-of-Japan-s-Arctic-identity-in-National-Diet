```python
from speechanalyzer import SpeechAnalyzer
from speechanalyzer import SpeechCharacteristics
from speechanalyzer import NetworkAnalyzer
import pandas as pd
from tqdm import tqdm
```

```python
with open ("stopwords Japanese.txt", encoding='utf-8') as f:
    stopwords = f.read().split('\n')
#df = pd.read_csv('北極_ND-speakers.csv', index_col = False)
df = pd.read_csv('hokkyoku_1949_2025.csv', index_col = False)
```

```python
corpus_analyzer = SpeechCharacteristics(df,1949,2025)
```

```python
speeches_all = corpus_analyzer.count_speech_number(figsize=(13,5))
speeches_all.to_csv('tables/speech_number_df_1949-2025.csv', index=False)
```

```python
speakers = corpus_analyzer.get_speakers()
speakers.to_csv('tables/speakers_df_1949-2025.csv', index=False)
speakers
```

```python
len(set(speakers.Speaker))
```

```python
invited_speakers = corpus_analyzer.get_share_of_guest_speakers()
invited_speakers.to_csv('tables/invited_speakers_1949-2025.csv', index=False)
invited_speakers
```

```python
party_membership = corpus_analyzer.get_party_shares(14,14)
party_membership.to_csv('tables/party_membership_1949-2025.csv', index=False)
party_membership
```
