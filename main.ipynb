```python
import pandas as pd

# Example DataFrame
df = pd.DataFrame({
    'speaker': ['A', 'B', 'C'],
    'Topic_0': [0.1, 0.4, 0.2],
    'Topic_1': [0.3, 0.2, 0.5],
    'Topic_2': [0.6, 0.1, 0.3]
})

# List of topic columns you're interested in
topic_cols = ['Topic_0', 'Topic_1', 'Topic_2']

# Get the row (e.g., second row)
row = df.iloc[1]  # row with speaker B

# Get the topic with the maximum value
dominant_topic = row[topic_cols].idxmax()
max_value = row[topic_cols].max()

print("Dominant topic:", dominant_topic)
print("Score:", max_value)
```

```python
# Example: counts of dominant topics per year
df = pd.DataFrame({
    'year': [2021, 2022, 2023],
    'Topic_0': [5, 8, 6],
    'Topic_1': [10, 7, 9],
    'Topic_2': [3, 5, 7]
})
```

```python
import matplotlib.pyplot as plt

# Set index to year (optional for grouped bar)
df_plot = df.set_index('year')

# Transpose for grouped bar plotting
df_plot.T.plot(kind='bar')

plt.title("Dominant Topic Counts per Year")
plt.xlabel("Topic")
plt.ylabel("Count")
plt.legend(title="Year")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Your nested dict
data = {
    1949: {'Topic_0': 1, 'Topic_1': 1, 'Topic_2': 1},
    1950: {'Topic_0': 0, 'Topic_1': 1, 'Topic_2': 1},
    1951: {'Topic_0': 2, 'Topic_1': 0, 'Topic_2': 0}
}

# Convert to DataFrame
df = pd.DataFrame.from_dict(data, orient='index')
df.index.name = 'year'
df = df.sort_index()  # Sort years if needed
df.plot(kind='bar')

plt.title("Dominant Topic Counts per Year")
plt.xlabel("Year")
plt.ylabel("Count")
plt.legend(title="Topic")
plt.tight_layout()
plt.show()
```

```python
df.plot(kind='bar', stacked=True)

plt.title("Dominant Topic Counts per Year (Stacked)")
plt.xlabel("Year")
plt.ylabel("Count")
plt.legend(title="Topic")
plt.tight_layout()
plt.show()
```
