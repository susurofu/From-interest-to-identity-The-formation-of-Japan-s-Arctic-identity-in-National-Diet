```python
from speechanalyzer import SpeechAnalyzer
from speechanalyzer import SpeechCharacteristics
from speechanalyzer import NetworkAnalyzer
from speechanalyzer import replace_jap_names
import pandas as pd
from tqdm import tqdm
```

```python
with open ("stopwords Japanese.txt", encoding='utf-8') as f:
    stopwords = f.read().split('\n')
df = pd.read_csv('北極-2025-09-30.csv', index_col = False)
```

```python
df = replace_jap_names(df)
```

```python
corpus_analyzer_1959 = SpeechCharacteristics(df,1959,1968)
speeches_1959_1968 = corpus_analyzer_1959.count_speech_number()
speeches_1959_1968.to_csv('tables/speech_number_df_1959-1968.csv', index=False)
```

```python
invited_speakers = corpus_analyzer_1959.get_share_of_guest_speakers()
invited_speakers.to_csv('tables/invited_speakers_1959-1968.csv', index=False)
invited_speakers
```
