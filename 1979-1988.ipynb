```python
import speechanalyzer
from speechanalyzer import SpeechAnalyzer
from speechanalyzer import SpeechCharacteristics
from speechanalyzer import NetworkAnalyzer
from speechanalyzer import replace_jap_names
import pandas as pd
from tqdm import tqdm
```

```python
with open ("stopwords Japanese.txt", encoding='utf-8') as f:
    stopwords = f.read().split('\n')
df = pd.read_csv('北極-2025-09-30.csv', index_col = False)
```

```python
df = replace_jap_names(df)
```

```python
corpus_analyzer_1979 = SpeechCharacteristics(df,1979,1988)
speeches_1979_1988 = corpus_analyzer_1979.count_speech_number()
speeches_1979_1988.to_csv('tables/speech_number_df_1979-1988.csv', index=False)
```

```python
invited_speakers = corpus_analyzer_1979.get_share_of_guest_speakers()
invited_speakers.to_csv('tables/invited_speakers_1979-1988.csv', index=False)
invited_speakers
```

```python
party_membership = corpus_analyzer_1979.get_party_shares(14,14)
party_membership.to_csv('tables/party_membership_1979-1988.csv', index=False)
party_membership
```
